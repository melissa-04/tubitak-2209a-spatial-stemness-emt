```python
import scipy.io as sio
import scipy.sparse as sp
import os

INP = '/content/04b_cytospace_inputs'
os.makedirs(INP, exist_ok=True)

# Matris gen × hücre yönüne transpoze (CytoSPACE bu yönü ister), CSR garanti
mat = adata_ref.X.T
mat = sp.csr_matrix(mat) if not sp.issparse(mat) else mat.tocsr()
sio.mmwrite(f'{INP}/scRNA_matrix.mtx', mat)

# Gen ve hücre isimleri, başlıksız, MTX ile aynı sırada
pd.Series(adata_ref.var_names).to_csv(f'{INP}/scRNA_genes.tsv',  sep='\t', index=False, header=False)
pd.Series(adata_ref.obs_names).to_csv(f'{INP}/scRNA_cells.tsv',  sep='\t', index=False, header=False)

# Hücre tipi etiketleri, başlıklı 2 sütun (CellType = celltype_major)
etiket = pd.DataFrame({
    'Cell IDs': adata_ref.obs_names,
    'CellType': adata_ref.obs['celltype_major'].values
})
etiket.to_csv(f'{INP}/cell_type_labels.txt', sep='\t', index=False)

print('Matris shape (gen × hücre):', mat.shape)
print('genes.tsv satır:', len(adata_ref.var_names))
print('cells.tsv satır:', len(adata_ref.obs_names))
print('etiket dosyası satır:', len(etiket))
print('\nÜretilen dosyalar:')
!ls -lh /content/04b_cytospace_inputs/
```

```python
import numpy as np

adata_st = sc.read_h5ad(f'{BASE}/03_spatial_analysis/visium_6patients_raw.h5ad')
comp = pd.read_csv(f'{BASE}/03_spatial_analysis/spatial_celltype_composition.csv', index_col=0)

print('Visium shape (spot × gen):', adata_st.shape)
print('obs kolonları:', list(adata_st.obs.columns))
print('\npatientid dağılımı:')
print(adata_st.obs['patientid'].value_counts())
print('\nanalyze=True spot sayısı:', int(adata_st.obs['analyze'].sum()))

print('\nKompozisyon CSV shape:', comp.shape)
print('Kompozisyon kolonları:', list(comp.columns))

# -ctfep tutarlılık testi: kompozisyondaki tipler == etiketteki 9 tip?
etiket_tipleri = set(adata_ref.obs['celltype_major'].unique())
comp_tipleri = set(c for c in comp.columns if c in etiket_tipleri)
print('\nEtiket tipleri (9):', sorted(etiket_tipleri))
print('Kompozisyonda eşleşen tip sayısı:', len(comp_tipleri), '/ 9')
```

```python
import scipy.sparse as sp

ST = '/content/04b_cytospace_inputs'
tipler = ['B-cells','CAFs','Cancer Epithelial','Endothelial','Myeloid',
          'Normal Epithelial','PVL','Plasmablasts','T-cells']
hastalar = ['1142243F','1160920F','CID4290','CID4465','CID44971','CID4535']

# Genler tüm hastalarda ortak → bir kez yaz
pd.Series(adata_st.var_names).to_csv(f'{ST}/ST_genes.tsv', sep='\t', index=False, header=False)

for h in hastalar:
    maske = (adata_st.obs['analyze'].values) & (adata_st.obs['patientid'].values == h)
    alt = adata_st[maske]
    spot_adlari = alt.obs_names

    # ST ekspresyonu: gen × spot yönüne transpoze, MTX yaz
    m = alt.X.T
    m = sp.csr_matrix(m) if not sp.issparse(m) else m.tocsr()
    sio.mmwrite(f'{ST}/{h}_ST_matrix.mtx', m)
    pd.Series(spot_adlari).to_csv(f'{ST}/{h}_ST_spots.tsv', sep='\t', index=False, header=False)

    # Koordinat: SpotID + array_row + array_col (piksel değil), başlıklı
    koord = pd.DataFrame({
        'SpotID': spot_adlari,
        'row': alt.obs['array_row'].values,
        'col': alt.obs['array_col'].values
    })
    koord.to_csv(f'{ST}/{h}_coords.tsv', sep='\t', index=False)

    # -ctfep oran dosyası: bu hastanın analyze spotlarının ortalama kompozisyonu, 1'e normalize
    comp_h = comp.loc[spot_adlari, tipler]
    ort = comp_h.mean(axis=0)
    ort = ort / ort.sum()
    frac = pd.DataFrame([ort.values], columns=tipler, index=['Fraction'])
    frac.to_csv(f'{ST}/{h}_cellfracs.txt', sep='\t')

    print(f'{h}: {maske.sum()} spot | matris {m.shape} (gen×spot) | oran toplamı {ort.sum():.4f}')

print('\nTüm hasta dosyaları üretildi.')
```

```python
import pandas as pd, os

hastalar = ['1142243F','1160920F','CID4290','CID4465','CID44971','CID4535']
OUTDIR = '/content/04b_cytospace_out'
epitel_tipler = ['Cancer Epithelial','Normal Epithelial']

# 1) 6 hastanın atamalarını birleştir
parcalar = []
for h in hastalar:
    df = pd.read_csv(f'{OUTDIR}/{h}/assigned_locations.csv')
    df['patientid'] = h
    parcalar.append(df)
tum = pd.concat(parcalar, ignore_index=True)
print('Toplam atanan hücre (6 hasta):', len(tum))
print('Toplam benzersiz spot:', tum['SpotID'].nunique())

# 2) Epitel atamalara CytoTRACE skoru bağla
epi = tum[tum['CellType'].isin(epitel_tipler)].copy()
epi['stem'] = epi['OriginalCID'].map(cyto['CytoTRACE2_Score'])
print('Epitel atama:', len(epi), '| skorlanan:', epi['stem'].notna().sum(),
      f"({epi['stem'].notna().mean()*100:.1f}%)")

# 3) Spot başına ortalama epitel skoru = spatial_stemness
stem = (epi.groupby('SpotID')
           .agg(spatial_stemness=('stem','mean'),
                n_epi_cells=('stem','size'),
                patientid=('patientid','first'))
           .reset_index())
print('spatial_stemness üretilen spot:', len(stem))
print(stem['spatial_stemness'].describe()[['mean','min','max']].round(3).to_dict())

# 4) DRIVE'A KAYDET (kritik — yerelden Drive'a)
SAVE = f'{BASE}/03_spatial_analysis'
os.makedirs(SAVE, exist_ok=True)
tum.to_csv(f'{SAVE}/cytospace_assignments_all.csv', index=False)
stem.to_csv(f'{SAVE}/spatial_stemness_scores.csv', index=False)

# 5) Doğrula: Drive'da dosyalar var mı?
for f in ['cytospace_assignments_all.csv','spatial_stemness_scores.csv']:
    p = f'{SAVE}/{f}'
    print(f'{os.path.exists(p)}  {os.path.getsize(p)//1024} KB  <-  {f}')
```

```python
import pandas as pd

stem = pd.read_csv(f'{BASE}/03_spatial_analysis/spatial_stemness_scores.csv')

# Visium obs'undan bölge etiketlerini bağla (SpotID = obs index)
obs = adata_st.obs[['patientid','subtype','region_class','tumor_region','analyze']].copy()
obs.index.name = 'SpotID'
stem = stem.merge(obs.reset_index()[['SpotID','subtype','region_class','tumor_region']],
                  on='SpotID', how='left')

print('spatial_stemness spot sayısı:', len(stem))
print('Bölge (tumor_region) dağılımı:')
print(stem['tumor_region'].value_counts(dropna=False))

print('\n--- Tutarlılık: hücre-düzeyi vs spot-düzeyi (aynı mertebede mi?) ---')
print('Hücre-düzeyi CytoTRACE (epitel):',
      f"medyan {cyto['CytoTRACE2_Score'].median():.3f}, ort {cyto['CytoTRACE2_Score'].mean():.3f}")
print('Spot-düzeyi spatial_stemness:   ',
      f"medyan {stem['spatial_stemness'].median():.3f}, ort {stem['spatial_stemness'].mean():.3f}")
```

```python
import numpy as np

# --- front vs core, hasta hasta ---
print('=== front vs core stemness (hasta hasta) ===')
print(f"{'Hasta':<10} {'front':>7} {'core':>7} {'fark':>7} {'yön':>6}  (n_f/n_c)")
yonler = []
for h in ['1142243F','1160920F','CID4290','CID4465','CID44971','CID4535']:
    d = stem[stem['patientid']==h]
    f = d[d['tumor_region']=='front']['spatial_stemness']
    c = d[d['tumor_region']=='core']['spatial_stemness']
    if len(f)>0 and len(c)>0:
        fark = f.mean()-c.mean()
        yon = 'front↑' if fark>0 else 'core↑'
        yonler.append(yon)
        print(f"{h:<10} {f.mean():>7.3f} {c.mean():>7.3f} {fark:>+7.3f} {yon:>6}  ({len(f)}/{len(c)})")

print(f"\nYön tutarlılığı: {yonler.count('front↑')}/{len(yonler)} hasta front↑")

# --- genel ---
gf = stem[stem['tumor_region']=='front']['spatial_stemness']
gc = stem[stem['tumor_region']=='core']['spatial_stemness']
print(f"\nGenel: front {gf.mean():.3f} vs core {gc.mean():.3f} (fark {gf.mean()-gc.mean():+.3f})")

# --- CID44971 altın maden testi ---
print('\n=== CID44971 kontrolü (altın maden: %51.7 oligopotent+ epitel) ===')
print('Hasta bazında spatial_stemness ortalaması:')
print(stem.groupby('patientid')['spatial_stemness'].mean().round(3).sort_values(ascending=False).to_string())
```
